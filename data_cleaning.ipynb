```python
df.columns = df.columns.str.strip().str.lower()
vendor_df.columns = vendor_df.columns.str.strip().str.lower()
plant_df.columns = plant_df.columns.str.strip().str.lower()

```

```python
def id_cleaned(data,id_columns):
  for col in id_columns:
    data[col]=(data[col].astype("string").str.strip())
  return data
```

```python
df=id_cleaned(df,["material_id","vendor_id","plant_id"])
vendor_df=id_cleaned(vendor_df,["vendor_id"])
plant_df=id_cleaned(plant_df,["plant_id"])
```

```python
text_column = [
    "material_description",
    "material_category",
    "sub_category",
    "manufacturer",
    "manufacturer_part_no",
    "uom",
    "vendor_id",
    "plant_id",
    "currency"
]

for col in text_column:

    if col in df.columns:

        df[col] = (
            df[col]
            .astype("string")
            .str.strip()
        )
```

```python
df[df["manufacturer"].isnull()]
```

```python
missing_values = [
    "",
    "NA",
    "N/A",
    "NULL",
    "null",
    "None",
    "none",
    "NaN",
    "nan"
]

df = df.replace(missing_values, pd.NA)
```

```python
df.isnull().sum()
```

```python
df[df["manufacturer"].isnull()]
```

```python
df["unit_price"] = pd.to_numeric(df["unit_price"],errors="coerce")

df["lead_time_days"] = pd.to_numeric(df["lead_time_days"],errors="coerce")
```

```python
df[df["unit_price"].isnull()]
```

```python
df["unit_price"].isnull().sum()
```

```python
df["created_date"]=pd.to_datetime(df["created_date"],errors="coerce")
df["last_updated"]=pd.to_datetime(df["last_updated"],errors="coerce")
```

```python
df["manufacturer"].head()
```

```python
df["manufacturer"] = (
    df["manufacturer"]
    .astype("string")
    .str.upper()
    .str.replace(".", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)

manufacturer_mapping = {

    "SKF INDIA": "SKF",
    "SKF INDIA LTD": "SKF",
    "SKF INDIA LIMITED": "SKF",
    "SKFINDA": "SKF",

    "ABB INDIA": "ABB",
    "ABB INDIA LTD": "ABB",
    "ABB INDIA LIMITED": "ABB",

    "BOSCH INDIA": "BOSCH",
    "BOSCH INDIA LTD": "BOSCH",
    "BOSCH INDIA LIMITED": "BOSCH"
}

df["manufacturer"] = df["manufacturer"].replace(
    manufacturer_mapping
)
```

```python
df.columns
```

```python
df['currency']=df['currency'].astype("string").str.strip()
```

```python
df['currency']=df['currency'].str.upper()
```

```python
valid_currencies = {
    "INR",
    "USD",
    "AED"
}

df["currency_valid"] = (
    df["currency"].isin(valid_currencies)
)
```

```python
df[df['currency_valid']==False]
```

```python
valid_uom = {
    "EA",
    "KG",
    "M",
    "LTR"
}

df["uom_valid"] = (
    df["uom"].isin(valid_uom)
)
```

```python
df["price_valid"] = ( df["unit_price"].notna()&(df["unit_price"] > 0))
df["lead_time_valid"] = (
    df["lead_time_days"].notna()&(df["lead_time_days"] >= 0)
)
```

```python
df["date_valid"] = df["created_date"].notna() & df["last_updated"].notna() & (df["last_updated"] >= df["created_date"])
```

```python
df["material_id"].duplicated().sum()
```

```python
df["material_id_unique"] = (
    ~df["material_id"].duplicated(keep=False)
)
```

```python
df[df["material_id_unique"] == False]
```

```python

valid_vendor_ids = set(
    vendor_df["vendor_id"]
    .dropna()
    .astype("string")
    .str.strip()
)

df["vendor_valid"] = (
    df["vendor_id"].notna()
    &
    df["vendor_id"].isin(valid_vendor_ids)
)
```

```python
valid_plant_ids = set(
    plant_df["plant_id"]
    .dropna()
    .astype("string")
    .str.strip()
)

df["plant_valid"] = (
    df["plant_id"].notna()
    &
    df["plant_id"].isin(valid_plant_ids)
)
```

```python

required_columns = [
    "material_description",
    "material_category",
    "sub_category",
    "manufacturer",
    "manufacturer_part_no",
    "uom",
    "vendor_id",
    "plant_id",
    "unit_price",
    "currency",
    "lead_time_days"
]

df["completeness_score"] = (
    df[required_columns]
    .notna()
    .sum(axis=1)
    / len(required_columns)
    * 100
).round(2)
```

```python
df[df['price_valid']==False]
```

```python
df["missing_manufacturer"] = (
    df["manufacturer"].isna()
)

df["missing_part_no"] = (
    df["manufacturer_part_no"].isna()
)

df["missing_uom"] = (
    df["uom"].isna()
)

df["missing_vendor"] = (
    df["vendor_id"].isna()
)
df["missing_lead_time"] = (
    df["lead_time_days"].isna()
)
```

```python
duplicate_materials = (
    df["material_id"].duplicated(keep=False).sum()
)
```

```python
duplicate_materials
```

```python
quality_checks = [
    "material_id_unique",
    "vendor_valid",
    "plant_valid",
    "uom_valid",
    "currency_valid",
    "price_valid",
    "lead_time_valid",
    "date_valid"
]

df["quality_score"] = (
    df[quality_checks]
    .sum(axis=1)
    / len(quality_checks)
    * 100
).round(2)
```

```python
df
```

```python
def quality_status(score):

    if score >= 90:
        return "Good"

    elif score >= 70:
        return "Needs Review"

    else:
        return "Critical"


df["quality_status"] = (
    df["quality_score"]
    .apply(quality_status)
)
```

```python
df[df['quality_status']=='Critical']
```

```python

clean_material_df = df.copy()


# Remove technical validation columns
technical_columns = [
    "currency_valid",
    "uom_valid",
    "price_valid",
    "lead_time_valid",
    "date_valid",
    "material_id_unique",
    "vendor_valid",
    "plant_valid"
]

clean_material_df = clean_material_df.drop(
    columns=technical_columns,
    errors="ignore"
)


# ============================================================
# 27. CLEAN VENDOR MASTER
# ============================================================

clean_vendor_df = vendor_df.copy()

for col in clean_vendor_df.columns:

    if clean_vendor_df[col].dtype == "object":

        clean_vendor_df[col] = (
            clean_vendor_df[col]
            .astype("string")
            .str.strip()
        )

clean_vendor_df = clean_vendor_df.replace(
    missing_values,
    pd.NA
)


# ============================================================
# 28. CLEAN PLANT MASTER
# ============================================================

clean_plant_df = plant_df.copy()

for col in clean_plant_df.columns:

    if clean_plant_df[col].dtype == "object":

        clean_plant_df[col] = (
            clean_plant_df[col]
            .astype("string")
            .str.strip()
        )

clean_plant_df = clean_plant_df.replace(
    missing_values,
    pd.NA
)
```

```python
clean_material_df.to_csv(
    "material_master_cleaned.csv",
    index=False
)

clean_vendor_df.to_csv(
    "vendor_master_cleaned.csv",
    index=False
)

clean_plant_df.to_csv(
    "plant_master_cleaned.csv",
    index=False
)
```
